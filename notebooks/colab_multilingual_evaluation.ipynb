```python
# Step 5: Display Generated Empirical Evaluation Results
import json
import pandas as pd

report_path = 'reports/pretrained_model_comparison.json'
if os.path.exists(report_path):
    with open(report_path, 'r', encoding='utf-8') as f:
        data = json.load(f)
    print('Evaluation Timestamp:', data.get('evaluation_timestamp'))
    print('GPU Used:', data.get('gpu_name'))
    print('\n=== EMPIRICAL EVALUATIONS SUMMARY ===')
    for key, eval_data in data.get('evaluations', {}).items():
        print(f"\n-- {key} --")
        print(f"Model: {eval_data.get('model_id')}")
        print(f"Test Samples: {eval_data.get('test_samples')}")
        print(f"Accuracy: {eval_data.get('accuracy'):.4f} | ROC-AUC: {eval_data.get('roc_auc'):.4f}")
        print(f"Binary Toxic Prec: {eval_data.get('binary_toxic', {}).get('precision'):.4f} | Recall: {eval_data.get('binary_toxic', {}).get('recall'):.4f} | F1: {eval_data.get('binary_toxic', {}).get('f1'):.4f}")
        print(f"Macro F1: {eval_data.get('macro_f1'):.4f}")
        print(f"Confusion Matrix: {eval_data.get('confusion_matrix')}")
        print(f"Latency: {eval_data.get('latency_ms_per_sample')} ms/sample")
else:
    print('Report not found. Ensure Step 4 completed successfully.')
```
